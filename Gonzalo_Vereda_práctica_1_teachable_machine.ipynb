inputImagen.addEventListener(

        const archivo =
            event.target.files[0];


        if (!archivo) {
            return;
        }


        // Comprobar que sea una imagen
        if (!archivo.type.startsWith("image/")) {

// ==========================================
// CLASIFICAR IMAGEN
// ==========================================

async function clasificarImagen() {

    if (!modeloCargado || !modelo) {

        mensaje.textContent =
            "Primero debes iniciar el clasificador.";

        return;
    }


    if (!clasificadorActivo) {

        mensaje.textContent =
            "El clasificador está detenido.";

        return;
    }


    if (!imagenSeleccionada) {

        mensaje.textContent =
            "Selecciona una imagen.";

        return;
    }


    estado.textContent =
        "Analizando imagen...";


    try {

        // Realizar predicción
        const predicciones =
            await modelo.predict(
                imagenSeleccionada
            );


        // Obtener la predicción con mayor probabilidad
        let mejorPrediccion =
            predicciones[0];


        for (
            let i = 1;
            i < predicciones.length;
            i++
        ) {

            if (
                predicciones[i].probability >
                mejorPrediccion.probability
            ) {

                mejorPrediccion =
                    predicciones[i];

            }

        }


        // Nombre de la categoría
        const nombreClase =
            mejorPrediccion.className;


        // Porcentaje
        const confianza =
            mejorPrediccion.probability * 100;


        // Mostrar resultado
        mostrarResultado(
            nombreClase,
            confianza
        );


    } catch (error) {

        console.error(
            "Error durante la clasificación:",
            error
        );


        estado.textContent =
            "Error durante la clasificación.";


        mensaje.textContent =
            "Ha ocurrido un error al analizar la imagen.";

    }

}


// ==========================================
// MOSTRAR RESULTADO
// ==========================================

function mostrarResultado(
    nombreClase,
    confianza
) {

    const porcentajeRedondeado =
        confianza.toFixed(2);


    estado.textContent =
        "Clasificación completada.";


    porcentaje.textContent =
        "Coincidencia: " +
        porcentajeRedondeado +
        "%";


    // Convertimos a minúsculas para
    // comparar las categorías
    const claseNormalizada =
        nombreClase
            .toLowerCase()
            .normalize("NFD")
            .replace(/[\u0300-\u036f]/g, "");


    // ======================================
    // MÓVIL
    // ======================================

    if (claseNormalizada === "movil") {

        resultadoTexto.textContent =
            "📱 Móvil";


        if (confianza >= PORCENTAJE_MINIMO) {

            mensaje.textContent =
                "¡La imagen parece ser un móvil!";

        } else {

            mensaje.textContent =
                "La imagen tiene una coincidencia baja con la categoría móvil.";

        }

    }


    // ======================================
    // RELOJ
    // ======================================

    else if (claseNormalizada === "reloj") {

        resultadoTexto.textContent =
            "⌚ Reloj";


        if (confianza >= PORCENTAJE_MINIMO) {

            mensaje.textContent =
                "¡La imagen parece ser un reloj!";

        } else {

            mensaje.textContent =
                "La imagen tiene una coincidencia baja con la categoría reloj.";

        }

    }


    // ======================================
    // OTRA CATEGORÍA
    // ======================================

    else {

        resultadoTexto.textContent =
            nombreClase;


        mensaje.textContent =
            "Se ha detectado la categoría " +
            nombreClase +
            ".";

    }


    // Guardar en historial
    guardarResultado(
        nombreClase,
        confianza
    );

}


// ==========================================
// GUARDAR RESULTADO
// ==========================================

function guardarResultado(
    nombreClase,
    confianza
) {

    const historialGuardado =
        JSON.parse(
            localStorage.getItem(
                "historialClasificador"
            )
        ) || [];


    const nuevoResultado = {

        clase: nombreClase,

        confianza:
            confianza.toFixed(2),

        fecha:
            new Date().toLocaleString("es-ES")

    };


    // Añadir al principio
    historialGuardado.unshift(
        nuevoResultado
    );


    // Máximo 20 resultados
    if (historialGuardado.length > 20) {

        historialGuardado.pop();

    }


    // Guardar
    localStorage.setItem(
        "historialClasificador",
        JSON.stringify(
            historialGuardado
        )
    );


    mostrarHistorial();

}


// ==========================================
// MOSTRAR HISTORIAL
// ==========================================

function mostrarHistorial() {

    const historialGuardado =
        JSON.parse(
            localStorage.getItem(
                "historialClasificador"
            )
        ) || [];


    // No hay resultados
    if (historialGuardado.length === 0) {

        historial.innerHTML =
            "<p>No hay resultados todavía.</p>";

        return;

    }


    historial.innerHTML = "";


    historialGuardado.forEach(
        function (resultado) {

            const elemento =
                document.createElement("div");


            elemento.classList.add(
                "historial-item"
            );


            elemento.innerHTML = `

                <div>

                    <div class="nombre">
                        ${resultado.clase}
                    </div>

                    <div class="fecha">
                        ${resultado.fecha}
                    </div>

                </div>

                <div class="confianza">
                    ${resultado.confianza}%
                </div>

            `;


            historial.appendChild(
                elemento
            );

        }
    );

}

// ==========================================

mostrarHistorial();

El código JavaScript sirve para hacer funcionar una aplicación que permite analizar imágenes mediante inteligencia artificial. Para ello, utiliza un modelo creado con Teachable Machine, que se carga cuando el usuario pulsa el botón «Iniciar». Una vez cargado, el programa permite seleccionar una imagen desde el ordenador y muestra una vista previa antes de analizarla.

Cuando se selecciona una imagen, el programa comprueba que sea un archivo válido y, si el clasificador está activo, la analiza automáticamente. El modelo devuelve varias posibles categorías y sus porcentajes de coincidencia. El código selecciona la categoría con el porcentaje más alto y muestra el resultado en pantalla. Si detecta un móvil o un reloj, muestra un mensaje específico y utiliza un porcentaje mínimo del 60 % para indicar si la coincidencia es suficiente o baja.

Además, el programa guarda los resultados en el historial utilizando localStorage, una función del navegador que permite conservar información aunque se cierre o se recargue la página. Cada resultado incluye el nombre de la categoría, el porcentaje de coincidencia y la fecha y hora del análisis. El historial muestra los 20 resultados más recientes y dispone de un botón para borrarlos después de pedir confirmación al usuario.

Por último, el código permite detener el clasificador cuando se desee, muestra mensajes para informar del estado de la aplicación y avisa si se produce algún error al cargar el modelo o analizar una imagen. En conjunto, JavaScript se encarga de controlar todas las acciones de la aplicación, desde cargar el modelo hasta mostrar y guardar los resultados.
