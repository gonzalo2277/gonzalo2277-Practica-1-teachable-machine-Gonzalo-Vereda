**¿Cómo se usa el modelo exportado? ¿Cómo se comporta?**

---



**Cómo se usa:** Teachable Machine exporta el modelo en un archivo `keras_model.h5` y un archivo `labels.txt` con los nombres de las clases (Móvil y Reloj). Para usarlo se carga el modelo con Keras, se abre la imagen, se recorta y redimensiona a 224x224 píxeles, se normalizan sus valores y se le pasa a `model.predict()`. El modelo devuelve una probabilidad para cada clase, y se elige la mayor. Esa probabilidad es la confianza.

**Cómo se comporta:**
- **Imagen clara:** el modelo acierta con seguridad cuando la imagen se parece a las que vio al entrenar.
- **Imagen ambigua:** el modelo al ser una imagen que puede confundirse, la confianza es menor o se reparte entre las dos clases.
- **Imagen sin categoría:** el modelo aunque la imagen no era ni un móvil ni un reloj, el modelo eligió una de las dos clases igualmente, porque solo conoce esas dos y no tiene una opción de "ninguna".

**Conclusión:** el modelo funciona bien con imágenes parecidas a su entrenamiento, pero una confianza alta no siempre significa que acierte. Para mejorarlo se podría añadir una clase "Otros" con imágenes variadas o fijar un umbral mínimo de confianza.*Escribe aquí tu respuesta.*
